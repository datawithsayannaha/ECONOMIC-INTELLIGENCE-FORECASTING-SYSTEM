# QQ Plot 

## Objective

QQ Plot was used to compare the distribution of selected numerical variables against a theoretical normal distribution.

This analysis helps determine whether variables satisfy normality assumptions required for several statistical techniques such as correlation analysis, hypothesis testing, ANOVA, and regression modeling.

# One Sample T-Test

## Objective

To determine whether the sample mean of a variable differs significantly from a hypothesized population mean.

This test helps evaluate whether observed averages are statistically meaningful or likely due to random variation.

---

### Null Hypothesis (H0)

Sample Mean = Hypothesized Mean

### Alternative Hypothesis (H1)

Sample Mean ≠ Hypothesized Mean

```python
variable = "Inflation_Rate"
sample = df_clean[variable].dropna()
```

```python
t_stat, p_value = ttest_1samp(sample,popmean = 5)
print("T Statistic : ", round(t_stat,4))
print("P value :", round(p_value,6))
```

```python
alpha = 0.05
if p_value < alpha:    
    print( "Reject Null Hypothesis")   
else:  
    print("Fail to Reject Null Hypothesis")
```

```python
test_config = {
    "Inflation_Rate": 5,
    "Interest_Rate": 3,
    "Unemployment_Rate": 6,
    "GDP_Growth_Rate": 0,
    "Composite_Risk_Score": 4
}

results = []

for var, hypothesized_mean in test_config.items():
    sample = df_clean[var].dropna()
    t_stat, p_value = ttest_1samp(sample,popmean=hypothesized_mean)
    decision = (
        "Reject H0"
        if p_value < 0.05
        else "Fail to Reject H0"
    )
    results.append([
        var,
        hypothesized_mean,
        round(sample.mean(),3),
        round(t_stat,3),
        p_value,
        decision
    ])

t_test_summary = pd.DataFrame(
    results,
    columns=[
        "Variable",
        "Hypothesized_Mean",
        "Sample_Mean",
        "T_Statistic",
        "P_Value",
        "Decision"
    ]
)
t_test_summary
```

# Confidence Interval (CI)

```python
variable = "Inflation_Rate"
sample = df_clean[variable].dropna()
n = len(sample)
mean = sample.mean()
std = sample.std()
print("Sample Size :", n)
print("Mean :", round(mean,4))
print("Std :", round(std,4))
```

```python
se = std / np.sqrt(n)
print("Standard Error :", round(se,6))
```

```python
confidence = 0.95
margin_error = t.ppf((1 + confidence) / 2,df=n-1) * se

lower_ci = mean - margin_error

upper_ci = mean + margin_error

print("Lower CI :", round(lower_ci,4))

print("Upper CI :", round(upper_ci,4))
```

```python
ci_summary = pd.DataFrame({
    "Variable":[variable],
    "Mean":[round(mean,4)],
    "Lower_CI":[round(lower_ci,4)],
    "Upper_CI":[round(upper_ci,4)]
})

ci_summary
```

- VIF < 5 → Acceptable
- VIF 5–10 → Moderate multicollinearity
- VIF > 10 → Severe multicollinearity

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

import numpy as np
```

```python
def evaluate_forecast(
    actual,
    predicted
):
    mae = mean_absolute_error(
        actual,
        predicted
    )
    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )
    mape = np.mean(
        np.abs((actual - predicted) / actual)) * 100
    return mae, rmse, mape
```

```python
hw_mae, hw_rmse, hw_mape = evaluate_forecast(
    test,
    hw_pred
)
arima_mae, arima_rmse, arima_mape = evaluate_forecast(
    test,
    arima_pred
)
sarima_mae, sarima_rmse, sarima_mape = evaluate_forecast(
    test,
    sarima_pred
)
```

```python
forecast_comparison = pd.DataFrame({

    "Model":[
        "Holt-Winters",
        "ARIMA",
        "SARIMA"
    ],
    "MAE":[
        hw_mae,
        arima_mae,
        sarima_mae
    ],
    "RMSE":[
        hw_rmse,
        arima_rmse,
        sarima_rmse
    ],
    "MAPE":[
        hw_mape,
        arima_mape,
        sarima_mape
    ]
})
forecast_comparison.round(4)
```

This project developed a complete **Economic Intelligence & Forecasting System** using economic, financial, demographic, and risk-related indicators across multiple countries and time periods.

The primary objective was to analyze economic conditions, identify key risk drivers, segment economic profiles, build predictive models, and forecast future economic risk levels using Machine Learning and Time Series Analytics.

---

## Data Preparation & Feature Engineering

The dataset was cleaned, transformed, and enriched through extensive feature engineering.

Several business-oriented indicators were created, including:

- Economic Health Index
- Purchasing Power Stress
- Debt Pressure
- External Vulnerability
- Trade Dependency
- Inflation-Interest Gap
- High Risk Flag

These engineered features helped capture complex economic relationships and improved model performance.

---

## Exploratory Data Analysis (EDA)

Comprehensive exploratory analysis was conducted to understand economic patterns and relationships.

Key analyses included:

- Distribution Analysis
- Correlation Analysis
- Outlier Detection
- Economic Trend Analysis
- Country and Regional Comparisons

These analyses helped uncover hidden patterns and identify major economic drivers.

---

## Machine Learning Analysis

Multiple Machine Learning techniques were applied to gain deeper insights into economic risk.

### Regression Models

The following regression models were developed and evaluated:

- Linear Regression
- Ridge Regression
- Lasso Regression
- Random Forest Regression

Among them, **Random Forest Regression** achieved the strongest predictive performance.

### Classification Analysis

Classification techniques were used to categorize observations into different economic risk levels.

This helped identify high-risk economic conditions and improve decision-making.

### Clustering Analysis

K-Means Clustering was applied to group economies with similar characteristics.

This revealed distinct economic segments and behavioral patterns.

### Explainable AI

Feature Importance and SHAP Explainability were used to understand model behavior.

The most influential economic risk drivers were:

- Purchasing Power Stress
- Government Debt (% GDP)
- Gini Coefficient
- Corruption Index
- Unemployment Rate

These techniques improved model transparency and business interpretability.

---

## Time Series Analytics & Forecasting

A complete time series forecasting framework was developed.

The analysis included:

- Trend Analysis
- Seasonality Analysis
- Rolling Mean & Rolling Standard Deviation
- Stationarity Testing (ADF & KPSS)
- Time Series Decomposition
- ACF Analysis
- PACF Analysis

These techniques helped understand long-term economic behavior and prepare the data for forecasting.

---

## Forecasting Models

Multiple forecasting approaches were developed and compared:

- Moving Average
- Simple Exponential Smoothing (SES)
- Holt Method
- Holt-Winters Method
- ARIMA
- SARIMA
- Prophet
- VAR (Vector Auto Regression)

Forecast accuracy was evaluated using:

- MAE
- RMSE
- MAPE

---

## Best Forecasting Model

Among all forecasting approaches, the **SARIMA model achieved the highest forecasting accuracy**.

### Forecast Accuracy Comparison

| Model | MAE | RMSE | MAPE |
|---------|---------|---------|---------|
| Holt-Winters | 0.0290 | 0.0350 | 0.7440 |
| ARIMA | 0.0348 | 0.0435 | 0.8967 |
| SARIMA | 0.0280 | 0.0337 | 0.7164 |

Based on these results, SARIMA was selected as the final forecasting model.

---

## Forecast Results

Using the SARIMA model:

- 3-Month Forecast was generated
- 6-Month Forecast was generated
- 12-Month Forecast was generated

The forecasts indicate that economic risk levels are expected to remain generally stable, with only minor seasonal fluctuations and no major risk shock anticipated.

This project successfully developed an end-to-end Economic Intelligence & Forecasting System by integrating Data Analytics, Machine Learning, Explainable AI, Clustering Techniques, and Advanced Time Series Forecasting.
